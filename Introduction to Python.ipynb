```python
print("YOUR FINAL GRADE")
print("Look at how you perfemed this semester.")

def grade_calc(english, maths, chemistry, physics, biology):
    average_score = (english + maths + chemistry + physics + biology) / 5

    if average_score >= 80:
        grade = "A"
    elif average_score >= 60:
        grade = "B"
    elif average_score >= 40:
        grade = "C"
    else: 
        grade = "D"
    return average_score, grade

english_score = (int(input("Enter your English Score: ")))
maths_score = int(input("Enter your Maths Score: "))
chemistry_score = int(input("Enter your Chemisty Score: "))
physics_score = int(input("Enter your Physics Score: "))
biology_score = int(input("Enter your Biology Score: "))

avg_score, grading = grade_calc(english_score, maths_score, chemistry_score, physics_score, biology_score)

print(f"""English = {english_score}
Mathematics = {maths_score}
Chemistry = {chemistry_score}
Physics = {physics_score}
Biology = {biology_score}""")
print(f"Your Average Grade is {avg_score}%")
print(f"Your Grade is {grading}.")
```

```python
print("PYTHON BASIC CALCULATOR")
def simple_calc(num1, num2, operation):
    if operation == "+":
        return num1 + num2
    elif operation == "-":
        return num1 - num2 
    elif operation == "*":
        return num1 * num2 
    elif operation == "/":
        if num2 == 0:
            print("Number cannot be divided by 0.")
        return num1 / num2
    else: return "Invalid Operation. Kindly select +, -, *, /"

first_number = int(input("Enter a number: "))
operation = input("Select an operation: +, -, *, /")
second_number = int(input("Enter a number: "))

result= simple_calc(first_number, second_number, operation)

print(f"{first_number} {operation} {second_number} = {result}")
```

```python
shopping_cart = {}

print("WELCOME TO YOUR SHOPPING CART")
print("Enter 'done' after inputting all items.")

while True:
    item = input("Add an item: ").strip().capitalize()

    if item.lower() == "done":
        break

    item_price = float(input("Add item price: "))

    shopping_cart[item] = item_price

total_cost = sum(shopping_cart.values())


for items, price in shopping_cart.items():
    print(f"{items} = N{price:,.2f}")

print(f"Your Total cost is N{total_cost}")

```

```python
import random

print("GUESS THE RIGHT NUMBER!")
print("You have to guess a number between 1 - 20.")

comp_random = random.randint(1, 20)

while True:
    randon_num = int(input("Select a random number from 1 - 20: "))
    if randon_num > 20:
        print("Select a number between 1 - 20. Try again!")
        break
    try:   
        if randon_num < comp_random:
            print("Higher!")
        elif randon_num > comp_random:
            print("Lower!")
        elif randon_num == comp_random:
            print(f"You got it! The number is {comp_random}.")
            break  
    except ValueError:
        print("Please enter a valid number.")
                       
```

4. Data Types in Python are classifications of data that a variable holds. It determins what type of operation can be performed on the data. Common examples includes:
- Numeric Types: categorized into 
    - integer - positive or negative whole numbers without decimals e.g 4, 55, -322
    - floating point - numbers with decimals e.g 34.55, 23
    - complex undefined numbers e.g x = 3 + 2s
- String - text data enclosed in single, double or triple quotes e.g "Alice", 'How are you?'
- Boolean - true or false
- List - series of mutable obeject in python written in square brackets [] e.g ["apples", 3.55, true]
- Tuple - series of immutable object seperated by commas in python written in parentheses () e.g (3, 5, 3)
- Dicionary - mutable collection of key-value pair written in curly bracket {} e.g {"name": "Jane Patrick"}
- Set - mutable ordered collection of items seperated by commas, written in curly braces {} e.g {1, 2, 33}
- NoneType - default state or empty variable

10. Python Dicionaries are mutable collections of key-value pair written in curly bracket {}. Example: 

`user_details = {"name": "Jane Patrick", "age": 35, "occupation" : "Mentalist"}`

In the above example "name", "age" and "occupation" represents the keys and the information passed represents the value of the key.

To access a value in a dictionary, the key is called in square brackets [], alongside the variable name. 

Eg. `print(user_details["name"]) returns "Jane Patrick"`. 

SInce dictionaris are mutable, a value can be modified and data can be added to the existing dictionary. 

Eg: `user_details["name"] = "Sheldon Cooper"` updates the existing dictionary "name"to Sheldon Cooper.

`user_details["best_friend"] = "Leonard"` adds a new key-value pair to user_details variable so that we now have, 

`user_details = {"name": "Sheldon Cooper", "age": 35, "occupation" : "Mentalist", "best_friend" : "Leonard"}`
